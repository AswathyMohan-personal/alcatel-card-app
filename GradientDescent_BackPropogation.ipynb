```python
import numpy as np

# Training Data
x = 2
y = 1

# Initial Parameters
w = 0.5
b = 0.1

learning_rate = 0.1

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def mse(actual, predicted):
    return (actual - predicted) ** 2

for epoch in range(1000):

    # Forward Pass
    z = w * x + b
    prediction = sigmoid(z)
    loss = mse(y, prediction)

    # Backpropagation
    dL_dy = -2 * (y - prediction)
    dy_dz = prediction * (1 - prediction)

    dz_dw = x
    dz_db = 1

    dL_dw = dL_dy * dy_dz * dz_dw
    dL_db = dL_dy * dy_dz * dz_db

    # Gradient Descent
    w = w - learning_rate * dL_dw
    b = b - learning_rate * dL_db

    if epoch % 100 == 0:
      print(
        f"Epoch {epoch}, "
        f"Loss = {loss:.6f}, "
        f"Prediction = {prediction:.6f}"
      )
```
